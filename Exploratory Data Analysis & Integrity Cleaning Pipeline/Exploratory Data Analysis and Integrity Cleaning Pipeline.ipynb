# Outlier Detection & Treatment
## 14. Outlier Count per Numerical Column (IQR Method)

```python
# Calculate the number of outliers using the IQR method

for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    
    print(f"{column}: {len(outliers)} outliers")
```

## 15. Inspect Outlier Transactions in 'Total'

```python
# Identify the transactions that are outliers in Total

Q1 = df["Total"].quantile(0.25)
Q3 = df["Total"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

total_outliers = df[
    (df["Total"] < lower_bound) | 
    (df["Total"] > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of outliers:", len(total_outliers))

total_outliers
```

# Missing Value Imputation
## 20. Systematic Imputation Pipeline (Median for Numerical, Mode for Categorical)

```python
# Systematic imputation pipeline: numeric -> median, categorical -> mode
# This makes the pipeline robust even though this dataset currently has 0 missing values

numerical_columns = df.select_dtypes(include=np.number).columns
categorical_columns = df.select_dtypes(include="object").columns

for column in numerical_columns:
    if df[column].isnull().sum() > 0:
        median_value = df[column].median()
        df[column] = df[column].fillna(median_value)
        print(f"Imputed '{column}' with median: {median_value}")

for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        mode_value = df[column].mode()[0]
        df[column] = df[column].fillna(mode_value)
        print(f"Imputed '{column}' with mode: {mode_value}")

print("\nMissing values remaining after imputation:")
print(df.isnull().sum().sum())
```

# Outlier Treatment
## 21. Cap Outliers Using IQR Bounds (Winsorization)

```python
# Treat outliers by capping (winsorizing) instead of dropping,
# since Tax 5%, Total, Cost of goods sold, and Gross income are all
# mathematically derived from Unit price x Quantity — dropping rows would
# distort other correlated fields.

columns_to_treat = ["Tax 5%", "Total", "Cost of goods sold", "Gross income"]

for column in columns_to_treat:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    before = df[(df[column] < lower_bound) | (df[column] > upper_bound)].shape[0]
    df[column] = np.where(df[column] < lower_bound, lower_bound, df[column])
    df[column] = np.where(df[column] > upper_bound, upper_bound, df[column])
    after = df[(df[column] < lower_bound) | (df[column] > upper_bound)].shape[0]

    print(f"{column}: {before} outliers capped -> {after} remaining")
```
